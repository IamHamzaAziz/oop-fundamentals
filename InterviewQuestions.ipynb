# 1. What is the Diamond Problem in Multiple Inheritance, and How Does Python Handle It?
**Explain the diamond problem that happens when a class inherits from two classes that both inherit from the same base class. How does Python fix this?**

The diamond problem occurs in multiple inheritance when a subclass inherits conflicting methods or attributes from a common ancestor through different paths, leading to ambiguity. In Python, it's resolved using the Method Resolution Order (MRO), which follows a C3 linearization algorithm to determine the order of inheritance.



```python
class A:
    def greet(self):
        return "Hello from A"

class B(A):
    def greet(self):
        return "Hello from B"

class C(A):
    def greet(self):
        return "Hello from C"

class D(B, C):  # Diamond: D -> B -> A and D -> C -> A
    pass

# Check MRO
print(D.__mro__)  # Output: (<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)

d = D()
print(d.greet())  # Output: Hello from B (follows MRO: B before C)
```

Interview Tip: Mention that changing inheritance order (e.g., D(C, B)) changes MRO to prioritize C, resolving differently.

To override or call super, use super() carefully

```python
class D(B, C):
    def greet(self):
        return super().greet() + " and D"

print(D().greet())  # Output: Hello from B and D (super() follows MRO)
```

# 2. What is the Difference Between Abstract Classes and Interfaces in Python?
**What's the difference between an abstract class and an interface? How do you make them in Python?**

Abstract classes can have both abstract (must-implement) and concrete methods, while interfaces (simulated in Python) only declare abstract methods without implementation. Python uses abc module for abstract classes, interfaces are just abstract classes with no concrete methods.

```python
from abc import ABC, abstractmethod

# Abstract Class (with concrete method)
class Shape(ABC):
    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass

    def describe(self):  # Concrete method
        return f"Shape with color {self.color}"

class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

# Interface-like (only abstract methods)
class Drawable(ABC):
    @abstractmethod
    def draw(self):
        pass

class Rectangle(Drawable):
    def draw(self):
        return "Drawing rectangle"

# Usage
circle = Circle("red", 5)
print(circle.area())       # 78.5
print(circle.describe())   # Shape with color red

rect = Rectangle()
print(rect.draw())         # Drawing rectangle
# Shape()                  # Error: Can't instantiate abstract class
```

# 3. Explain Method Overloading vs. Method Overriding in Python.
**What's the difference between method overloading (same name, different parameters) and overriding (redefining a parent method in child)?**

Overloading allows multiple methods with the same name but different signatures (not natively supported in Python; use defaults or *args). Overriding reimplements a parent method in a subclass for polymorphism.

```python
class Calculator:
    def add(self, a, b=None):  # Simulated overloading with defaults
        if b is None:
            return a  # Single arg
        return a + b

class AdvancedCalculator(Calculator):
    def add(self, a, b=None):  # Overriding
        if b is None:
            return super().add(a) * 2  # Override behavior
        return super().add(a, b) + 10  # Override with extra

# Usage
calc = Calculator()
print(calc.add(5))       # 5
print(calc.add(5, 3))    # 8 (int)
print(calc.add("Hi", " there"))  # Hi there (string)

adv = AdvancedCalculator()
print(adv.add(5))        # 10 (overridden)
print(adv.add(5, 3))     # 18 (overridden)
```
