# Classes & Objects

```python
class Person:
    will_have = "name"   # class attribute (shared by all Persons)

    # Constructor
    def __init__(self, name, age = 18):
        self.age = age
        self.name = name

    def hello(self):
        return f"Hello my name is {self.name} and I'm {self.age} years old"
```

## Making objects (instances)

```python
p1 = Person("Kabir")
p2 = Person("Naseer", 20)

print(p1.will_have)
```

## Using "is" with objects

```python
print(p1 is p2) # False (two different objects)
```

## Default attributes

```python
print(p1.age)   # Default age which is set to 18
print(p2.age)   # 20 which was set while this object was made
```

## Hello message with name and age

```python
print(p1.hello())
```

# Encapsulation

```python
class BankAccount:
    def __init__(self, owner, name, balance = 0):
        self.owner = owner
        self._name = name   # Protected (The single underscore is a signal to other developers: This is intended for internal use.)
        self.__balance = balance    # Private attribute, cannot be used outside class

    def getBalance(self):
        return self.__balance
```

```python
first_account = BankAccount('self', 'John', 20)
print(first_account.getBalance())
print(first_account._name)  # Can be accessed outside of class but it is not recommended
```

## Private Attribute Error
__balance cannot be accessed outside of class as it is private and if such thing is tried, an error is thrown just like below

```python
print(first_account.__balance)
```

## Getters & Setters
In Python @property simplifies getter/setter logic.

```python
class Member:
    def __init__(self, name, member_id):
        self._name = name           # Protected attribute
        self.__member_id = member_id  # Private attribute
        self._borrowed_books = []

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        if isinstance(new_name, str) and new_name:
            self._name = new_name
        else:
            raise ValueError("Name must be a non-empty string")

# Usage
member = Member("Alice", "M001")
print(member.name)               # Output: Alice
member.name = "Bob"             # Using setter
print(member.name)              # Output: Bob
```

# Methods

## Instance Method

```python
class Counter:
    def __init__(self):
        self.value = 0

    def inc(self):            # instance method which increments value
        self.value += 1

c = Counter()
c.inc()
print(c.value)  # 1
```

## Class Method
Often used for alternative constructors or class-level data

```python
class User:
    count = 0

    def __init__(self, username):
        self.username = username
        User.count += 1

    @classmethod
    def from_email(cls, email):
        username = email.split("@")[0]
        return cls(username)         # calls __init__
```

Here from_email function can be called while this object is being made or created

```python
u = User.from_email("hamza@example.com")
print(u.username)  # hamza
print(User.count)  # 1
```

## Static Method
No self, no cls. Just a helper function inside the class

```python
class MathTools:
    @staticmethod
    def is_even(n):
        return n % 2 == 0
```

```python
print(MathTools.is_even(4))  # True
```

# Inheritance

## Single Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return "..."

class Dog(Animal):                 # Dog IS-A Animal
    def speak(self):               # override method
        return "Woof"

d = Dog("Buddy")
print(d.name, d.speak())  # Buddy Woof
```

Use super() to call parent behavior

```python
class Cat(Animal):
    def __init__(self, name, lives=9):
        super().__init__(name)     # call Animal.__init__
        self.lives = lives
```

## Multilevel Inheritance
Chain of parent → child → grandchild.

```python
class Vehicle:
    def important(self):
      return "each vehicle should have a number"

    def move(self):
        return "moving"

class Car(Vehicle):
    def move(self):
        return "car moving"

class SportsCar(Car):
    def move(self):
        return "sports car moving fast"
```

```python
s = SportsCar()
print(s.move())  # sports car moving fast
print(s.important())  # getting method from main or top most class (in inheritance heirarchy)
```

## Multiple Inheritance
A class has two or more parents.

```python
class CameraMixin:
    def take_photo(self):
        return "click!"

class Phone:
    def call(self, number):
        return f"calling {number}"

class SmartPhone(Phone, CameraMixin):  # inherits both
    pass
```

```python
sp = SmartPhone()
print(sp.call("123"))
print(sp.take_photo())
```

## MRO (Method Resolution Order)
Python looks for methods from left to right in the bases, using C3 linearization. Use super() to be safe.

```python
class A:
    def who(self): return "A"

class B(A):
    def who(self):
        return "B -> " + super().who()

class C(A):
    def who(self):
        return "C -> " + super().who()

class D(B, C):
    def who(self):
        return "D -> " + super().who()
```

```python
d = D()
print(d.who())         # D -> B -> C -> A
print(D.mro())         # shows the search order
```

# Method overriding
Child class replaces a method from parent.
You can also call parent code with super().

```python
class FileSaver:
    def save(self, data):
        print("Saving to file:", data)

class SafeFileSaver(FileSaver):
    def save(self, data):              # override
        if data is None:
            raise ValueError("no data")
        print("Check passed")
        super().save(data)             # also calling parent
```

```python
s = SafeFileSaver()
s.save("hello")
```

We can override __init__ too

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)   # call parent __init__
        self.breed = breed

d = Dog("Buddy", "Golden Retriever")
print(d.name, "which is", d.breed)
```

# Abstraction
Make a base class with abstract methods. You cannot make objects from it. Child classes must implement those methods.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
  def __init__(self, sides):
      self.sides = sides

  @abstractmethod
  def area(self):
      pass

class Circle(Shape):
  def __init__(self, r):
    super().__init__(0)
    self.r = r
  def area(self): return math.pi * self.r * self.r

class Rectangle(Shape):
  def __init__(self, w, h):
    super().__init__(4)
    self.w, self.h = w, h
  def area(self): return self.w * self.h
```

```python
c = Circle(2)
r = Rectangle(3, 4)
print(c.area(), r.area())
print(c.sides, r.sides)
```

Error as an abstract class cannot be used to create an object directly

```python
s = Shape(5)
```

We are using ABC in class Shape(ABC) because:
- Sets a contract (e.g., every Shape must have area()).
- Helps us catch missing methods early.

# Magic / Dunder Methods
These are special methods that start and end with __.
<br>They let our class work with Python features (print, len, +, ==, sort, etc.)

```python
class User:
    def __init__(self, name): self.name = name
    def __repr__(self):       return f"User(name={self.name!r})"  # for devs/debug
    def __str__(self):        return self.name                    # for users

u = User("Hamza")
print(u)        # uses __str__  -> Hamza
print([u])      # list shows __repr__ -> [User(name='Hamza')]
```

```python
class Vector:
    def __init__(self, *nums):
        self.data = list(nums)

    def __repr__(self):
        return f"Vector({', '.join(map(str, self.data))})"

    def __len__(self):
        return len(self.data)

    def __add__(self, other):
        if len(self) != len(other):
            raise ValueError("size must match")
        return Vector(*[a+b for a,b in zip(self.data, other.data)])

    def __eq__(self, other):
        return self.data == other.data

    def __lt__(self, other):       # so we can sort
        return sum(self.data) < sum(other.data)

v1 = Vector(1,2,3)
v2 = Vector(3,1,0)
print(len(v1))        # 3
print(v1 + v2)        # Vector(4, 3, 3)
print(v1 == v2)       # False
print(sorted([v2, v1]))  # uses __lt__
```

# Polymorphism
Polymorphism allows different classes to be treated as instances of the same parent class, typically through method overriding or interface implementation.
- Polymorphism enables flexibility by allowing subclasses to provide specific implementations of a parent class’s method.
- In Python, polymorphism is achieved through method overriding and duck typing.

```python
class LibraryItem:
    def __init__(self, title, id):
        self.title = title
        self.id = id

    def check_out(self):
        return f"Checking out {self.title}"

class Book(LibraryItem):
    def check_out(self):  # Override
        return f"Book '{self.title}' checked out."

class DVD(LibraryItem):
    def check_out(self):  # Override
        return f"DVD '{self.title}' checked out."

# Polymorphic behavior
items = [Book("1984", "B001"), DVD("Inception", "D001")]
for item in items:
    print(item.check_out())
```
